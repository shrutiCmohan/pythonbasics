6. Why are lists mutable, but tuples are immutable?
    - Lists and Tuples are very important data structures while handling data. But there is difference in both.

7. What is the difference between “==” and “is” operators in Python?
    - Operators are the special keywords used to perform special operations on the values.
    - 'is' operator is known as identity operator. This checks if two variables point to the same object in the memory. It returns output as a TRUE if both variables are reffering to the exact same value and FALSE if not.
    - '==' operator is called as equality operator. When two variables have the same value this operator return TRUE otherwise FALSE.

8. What are logical operators in Python?
   - Logical operators are used to perform logical operations and returns boolean value. It combines conditional statement and perform operations based on given conditions. There are three types of logical operators=
   AND, OR, NOT
   - AND = It perform logical 'and' operation and return TRUE if both operands are true and returns FALSE if one of them is false.
   - OR = It perform logical 'or' operation and return TRUE value if atleast one operand is true otherwise returns FALSE if both of them are false.
   - NOT= It returns opposite boolean value. if operand is true it returns false and vice versa.

            item = [1,2,3,4]

               print (item)

```python
#14.Write a program to reverse a string using a while loop.
name= str(input("enter the name="))
i=len(name)-1
while i>=0:
    print(name[i],end='')
    i=i-1
```

```python
#15.Write a Python program to calculate the factorial of a number provided by the user using a while loop.
num= int(input("enter number="))
i=1
while num>0:
  i=i*num
  num=num-1
print("factorial of given number is ",i)
```
